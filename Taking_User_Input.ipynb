## The Prompt

A **prompt** is a message displayed to the user that tells them what information they should enter.

### A prompt should usually:



*   Clearly tell the user what to enter
*   End with a colon (:)
*   Include a space after the colon


For example:

```python
input("Enter your name: ")

The text "Enter your name: " is the prompt.
The Result from it would be like "Enter your name: Awais"
The space after : makes the interaction easier to read.

````python
input() # It has no prompt so just a blank box will appear with no message telling user what to enter creating confusions.
input("Whats is you name: ") # "What is your name: " is a prompt.
# When this code cell will run you will first see a completely blank box with no instruction. After you enter something, you will see a second box with a clear message at right side (called prompt)

# We can also store user inout in a variable so that later we can use them easily.
x=input("What is your name: ")
print(x)
````

## Understanding the `input()` Documentation

If you look at Python's documentation or inspect the function, you may see a signature similar to:

```text

  1.  Prompt is the parameter that allows us to provide a message to the user.
  2.  Object indicates that the prompt can be an object. For beginners, it is enough to understand that Python can accept a value here and display its text representation.
  3.  This means the default value of prompt is an empty string. That is why we leave input() empty nothing appear with the blank box.
  4.  The / means that prompt is positional-only. So normally we write input("Name: ") instead of input(prompt="Name: ").
  5.  This is the most important part for our lesson. It tells us that input() returns a value of type str — a string. This means that even if the user enters a number, input() initially gives us that value as a string. so for example if user input was 9, it will restur to us as "9".


````python
input() # As we had not write anything so the defualt value of prompt parameter would be applied on it and nothing will appear with the blank box.

input("Name: ") # We write it instead of input(prompt="Name: "). Becuase of /, there is no need to wrte prompt=

type(input("Age: ")) # type() is a function used to find the data type of something. You would see str at the end indicating that the result of input function is always in the string.
````

# Type Conversion

The solution is **type conversion**.

**Type conversion** means changing a value from one data type to another compatible data type.

For example:

```python
int(5.0)

## Types of Type conversion:

### Explicit Type Conversion:
Python automatically converts a value when necessary and when the conversion is safe. e.g 2 + 7.5 = 9.5 (here first 2 was converted into 2.0)

### Implicit Type Conversion:
Sometimes Python does not automatically convert values in the way we want. In those situations, we explicitly tell Python which type we want by using a conversion function. It is commenly kown as "Type Casting".

Python provides several built-in functions for common conversions, including:


*   int() → converts to an integer when possible
*   float() → converts to a floating-point number when possible
*   str() → converts a value to a string representation
*   complex() → converts a value to a complex number when possible
*   bool() → converts a value to a Boolean value

Other conversion functions also exist for collections, such as list().

#### Note:
Not every value can be converted into every data type. For example, int("23") can be converted into integer but int("hello") can't be converted into integer


````python
print(int("2"))
print(float(2))
print(bool(12))
print(complex(20))
print(str(2))

# We use print function with each one separately becuase if we dont use them only the result of last line would be shown and the others would not be shown.
# For type conversion, the original data type and desired data type should be compatibe with each other. if they are not compaitible, your code will crash.
````

````python
# Type conversion doesn't change or modify original value instead it create a new value.
a=3.5
print(int(a))
print(a) # The result of this line would be 3.5 (original one), proving that type() doesn't change the original value.
````

## Converting User Input into a Number

Now we can solve our original problem.

Since `input()` always returns a string, we can convert the user's input into an integer using `int()`.

We can do this in two ways.

### Method 1: Convert immediately

We can place `input()` inside `int()`:

```python
first_number = int(input("Enter the first number: "))

## Another Way: Convert the Values Later

We can also take the input first and convert it when performing the calculation.

For example:

```python
first_number = input("Enter the first number: ")
second_number = input("Enter the second number: ")

result = int(first_number) + int(second_number))

````python
# So the final code would be:

# Take two integer inputs from the user
first_number = int(input("Enter the first number: "))
second_number = int(input("Enter the second number: "))

# Add the two numbers
result = first_number + second_number

# Display the result
print("The sum is:", result)
````

````python
# Another way to write this code is:

# Take inputs from the user
first_number = int(input("Enter the first number: "))
second_number = int(input("Enter the second number: "))

# COnvert the two numbers into integer and Add the two numbers
result = int(first_number) + int(second_number)

# Display the result
print("The sum is:", result)
````
