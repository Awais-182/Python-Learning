### 🔹 2. Identifiers

* **Allowed Characters:** Can contain numbers, but **no spaces or special symbols** (`@`, `$`, `%`, etc.).
* **Word Separators:** Must use underscores (`_`) instead of spaces (`snake_case`).

### 🔹 What is a Literal?
A **literal** is raw data assigned directly to an identifier in a variable. Python features **4 primary literal types**:
1. **Numeric Literals**
2. **String Literals**
3. **Boolean Literals**
4. **Special Literal (`None`)**

#### 1. Binary Literals (`0b` / `0B`)
- Transistors in CPU chips operate in two states: `0` (OFF, no electricity) and `1` (ON, electricity flowing). This is the **Binary System**.
- Python assumes base-10 (Decimal) by default. To specify Binary, prefix the value with `0b` or `0B`.
- **Note:** Output is automatically evaluated and displayed as a decimal integer.

#### 2. Hexadecimal Literals (`0x` / `0X`)
- Base-16 number system. Prefix values with `0x` or `0X`. Always displays as a decimal integer.

#### 3. Octal Literals (`0o` / `0O`)
- Base-8 number system. Prefix values with `0o` or `0O`. Always displays as a decimal integer.

#### 4. Float Literals & Scientific Notation
- Stores decimal numbers (e.g., `1.3`, `4.5`).
- For very large or very small numbers, use **Scientific Notation** (`e` or `E` representing $10^x$). Example: `2.0e200` $= 2.0 \times 10^{200}$.
- **Key Rule:** Any number written in scientific notation defaults to the **`float` data type**.

#### 5. Integer / Decimal Literals
- Whole numbers (`1`, `100`). Leading zeros are invalid (e.g., `0123` throws an error).
- **Visual Separators:** Use underscores (`_`) instead of commas for large numbers (`1_000_000`).

#### 6. Complex Literals
- Complex numbers include real and imaginary components (`a + bj`).
- Uses `j` or `J` for the imaginary component.
- **Arithmetic Rules:**
  - Missing real part defaults to `0` (e.g., `3j` is `0 + 3j`).
  - Zero imaginary components result in `0j` (e.g., `(1+3j) - (2+3j)` $= -1 + 0j$).
  - Extract parts using `.real` and `.imag` attributes (returns floats).

````python
# Binary, Hexadecimal, and Octal Literals
binary_val = 0b1010    # 10 in binary
hex_val = 0xFF         # 255 in hex
octal_val = 0o12       # 10 in octal

print("Binary 0b1010:", binary_val)
print("Hex 0xFF:", hex_val)
print("Octal 0o12:", octal_val)

# Integers with Underscores
large_val = 1_000_000
print("\nReadable Integer:", large_val)

# Float Scientific Notation
sci_num = 2.0e5  # 2.0 * 10^5 = 200000.0
print("Scientific Notation Value:", sci_num)
print("Type of Scientific Notation:", type(sci_num))

# Complex Literals & Extraction
c1 = 1 + 3j
c2 = 2 + 3j
res1 = c1 - c2  # Result: -1 + 0j
res2 = (-1 + 3j) + (1 + 3j) # Result: 6j

print("\nComplex Calc 1:", res1)
print("Complex Calc 2:", res2)
print("Real Part of c1:", c1.real)
print("Imaginary Part of c1:", c1.imag)
````

#### 1. Single, Double & Multi-line Strings
- Enclose strings in `'...'` or `"..."`.
- Multi-line strings require triple quotes `"""..."""` or `'''...'''`.

#### 2. Unicode Strings
- Early computing only supported ASCII (English letters/numbers). As computing went global, **Unicode** was created to map every character, symbol, and emoji across human languages to a unique **Code Point**.
- **Usage:**
  - Directly typing the character inside quotes (e.g., `"اردو"`).
  - Using Unicode escape codes prefixed with `\u` (4 hex digits) or `\U` (8 hex digits).

#### 3. Escape Characters vs. Raw Strings
- **Escape Character (`\`):** Alters default text processing (e.g., `\n` for newline, `\t` for tab).
- **Raw String (`r"..."` or `R"..."`):** Disables escape character processing completely, treating backslashes as literal text (essential for file paths and regex).

````python
# Multi-line String
multi_str = """This is line 1
This is line 2"""
print(multi_str)

# Unicode Characters (Direct & Escape Codes)
urdu_text = "اردو"
unicode_char = "\u0905"  # Devanagari Letter A
emoji_symbol = "\U0001F600" # Grinning Face Emoji

print("\nDirect Unicode:", urdu_text)
print("Unicode Escape Code \\u0905:", unicode_char)
print("Unicode Emoji \\U0001F600:", emoji_symbol)

# Escape Sequences vs. Raw Strings
normal_path = "C:\\new_folder\\test"
raw_path = r"C:\new_folder\test"

print("\nNormal Path with Escapes:\n", "Line1\nLine2\t(Tabbed)")
print("Raw String Path:", raw_path)
````

- Standard truth states: `True` and `False` (capitalization is mandatory).
- **Under the Hood:** Python treats `True` as integer `1` and `False` as integer `0`.
- **Truthiness Rules (`bool()`):**
  - Almost everything evaluates to `True` if it has content.
  - Evaluates to `False` **only** if empty (e.g., `""`, `0`, `0.0`, `None`, empty parenthesis/brackets).

````python
# Boolean Arithmetic
print("True + 1 =", True + 1)      # 1 + 1 = 2
print("False + 1 =", False + 1)    # 0 + 1 = 1
print("True * 50 =", True * 50)    # 1 * 50 = 50

# Type Conversions with bool()
print("\nbool('Text'):", bool("Awais"))
print("bool(100):", bool(100))
print("bool(0):", bool(0))
print("bool(''):", bool(""))
````

- **`None`** represents the null state, emptiness, or absence of a value (must start with capital `N`, no quotes).
- **Comparison to C/C++:** In C/C++, you can declare an uninitialized variable without assigning a value. In Python, variables **cannot be uninitialized**. You use `None` as a explicit placeholder.
- **Example Use Case:** Storing a `highest_score` variable before a game starts. Setting it to `0` would be inaccurate (since `0` is a valid score), so it starts as `None`.

````python
# Placeholder initialized before actual value exists
highest_score = None

print("Initial Highest Score:", highest_score)
print("Type of None:", type(highest_score))

# Simulated Game Logic
player_score = 45

if highest_score is None or player_score > highest_score:
    highest_score = player_score

print("Updated Highest Score:", highest_score)
````

### 🔹 What is a Variable?
A variable is a human-friendly **label or nickname** attached to a value stored at a specific physical RAM address.

### 🔹 Components of a Variable
$$\text{Variable} = \text{Identifier} + \text{Assignment Operator } (=) + \text{Literal}$$
- **Identifier:** Label name (e.g., `Name`).
- **Assignment Operator (`=`):** Binds the identifier to the memory location of the literal.
- **Literal:** The underlying value stored in RAM (e.g., `"Awais"`).

### 🔹 RAM Allocation & `id()` Function
- Use `id(variable)` to inspect the memory address assigned to an object in RAM.
- **Garbage Collector (GC):** Python automatically releases RAM memory occupied by objects that no longer have active variable references.

### 🔹 Case Sensitivity
Identifiers are strictly case-sensitive. `age`, `Age`, `AGe`, and `AGE` are distinct references in RAM.

````python
# Variable structure
Name = "Awais"

# Memory address in RAM
print("Value:", Name)
print("RAM Address (id):", id(Name))

# Case Sensitivity Test
age = 20
Age = 25
AGE = 30

print(f"\nCase Sensitivity: age={age}, Age={Age}, AGE={AGE}")
print("Are RAM IDs different?", id(age) != id(Age) != id(AGE))
````

### 🔹 Three Ways to Assign Multiple Variables
1. **Semicolon Separation:** `a=1; b=2; c=3`
2. **Tuple Unpacking Style:** `a, b, c = 1, 2, 3`
3. **Chained Assignment:** `a = b = c = 6` (All identifiers point to the *exact same memory address*).

### 🔹 Increment and Decrement Operators
- Python **does not support** C-style unary operators like `++a` or `a++`.
- Use **Compound Assignment Operators**:
  - **Increment:** `a += 1` (equivalent to `a = a + 1`)
  - **Decrement:** `a -= 1` (equivalent to `a = a - 1`)

````python
# 1. Semicolon Style
x = 10; y = 20; z = 30
print("Semicolon style:", x, y, z)

# 2. Parallel Assignment
a, b, c = 1, 2, 3
print("Parallel style:", a, b, c)

# 3. Chained Assignment (Shared RAM address)
p = q = r = 500
print("Chained style:", p, q, r)
print("Shared RAM ID Check:", id(p) == id(q) == id(r))

# Increment and Decrement
count = 5
count += 2  # Increment by 2
print("\nIncremented Count (count += 2):", count)

count -= 1  # Decrement by 1
print("Decremented Count (count -= 1):", count)
````

### 🔹 Static Typing (C, Java)
- Variable data types must be declared explicitly before assignment (e.g., `int a = 1;`).
- You **cannot change the data type** of the variable later in execution.

### 🔹 Dynamic Typing (Python)
- Data types are detected automatically at runtime based on the assigned literal.
- You can freely rebind an identifier to a value of a completely different data type without errors.
- **Re-assignment behavior:** When you reassign a variable (e.g., `name = "Awais"` to `name = 100`), Python drops the old pointer and binds the variable name to the new object in RAM.

````python
# Variable initially references a String
data = "Awais"
print("Value:", data, "| Type:", type(data), "| RAM ID:", id(data))

# Re-assigned to an Integer
data = 100
print("Value:", data, "| Type:", type(data), "| RAM ID:", id(data))

# Re-assigned to a Float
data = 1.56
print("Value:", data, "| Type:", type(data), "| RAM ID:", id(data))
````
