# Comprehensive Guide to Python Operators

**Operators** are special symbols or keywords used to perform specific operations on values and variables.

* **Operand:** The data or value on which the operator acts.
* **Operator:** The symbol performing the operation.
* **Expression:** A combination of operands and operators that evaluates to a new value (e.g., in `5 + 2`, `5` and `2` are operands, `+` is the operator, and `5 + 2` is the expression).

---

## 1. Unary vs. Binary Operators

| Operator Category | Operands Required | Evaluation Direction | Examples |
| :--- | :--- | :--- | :--- |
| **Unary** | 1 operand | Right-to-Left (chained) | `+5`, `-7`, `not True`, `~a` |
| **Binary** | 2 operands | Left-to-Right | `5 + 2`, `a and b`, `x == y` |

> **Precedence Note:** Unary operators take higher priority than binary operators. In `10 + -5 * -2`, Python first evaluates the unary minuses to produce `10 + (-5) * (-2)`, then evaluates multiplication, and finally addition.

````python
# Unary operators (Right-to-Left evaluation)
print(- -5)       # Evaluates to: -(-5) -> 5
print(+ - -3)     # Evaluates to: +(-(-3)) -> 3

# Precedence: Unary evaluated before Binary
# 10 + (-5) * (-2) -> 10 + 10 -> 20
result = 10 + -5 * -2
print("10 + -5 * -2 =", result)
````

## 2. Arithmetic Operators
Arithmetic operators perform mathematical calculations. All are binary except `+` and `-`, which can act as unary operators.

| Operator | Name | Description | Example |
| :--- | :--- | :--- | :--- |
| `+` | Addition | Adds two values | `5 + 2 = 7` |
| `-` | Subtraction | Subtracts right from left | `5 - 2 = 3` |
| `*` | Multiplication | Multiplies two values | `5 * 2 = 10` |
| `/` | True Division | Divides and returns a float | `5 / 2 = 2.5` |
| `//` | Floor Division | Divides and rounds down to nearest integer | `5 // 2 = 2` |
| `%` | Modulus | Returns the division remainder | `10 % 3 = 1` |
| `**` | Exponentiation | Raises left operand to the power of right | `2 ** 3 = 8` |

````python
a = 10
b = 3

print("Addition (+):", a + b)
print("Subtraction (-):", a - b)
print("Multiplication (*):", a * b)
print("True Division (/):", a / b)
print("Floor Division (//):", a // b)
print("Modulus (%):", a % b)
print("Exponentiation (**):", a ** b)
````

## 3. Comparison (Relational) Operators
Comparison operators evaluate relationships between values and **always return a Boolean (`True` or `False`)**.

| Operator | Meaning | Example (`x=5, y=3`) | Result |
| :--- | :--- | :--- | :--- |
| `==` | Equal to | `5 == 3` | `False` |
| `!=` | Not equal to | `5 != 3` | `True` |
| `>` | Greater than | `5 > 3` | `True` |
| `<` | Less than | `5 < 3` | `False` |
| `>=` | Greater than or equal to | `5 >= 5` | `True` |
| `<=` | Less than or equal to | `3 <= 5` | `True` |

> **Note:** A single `=` is the assignment operator, not a comparison operator.

````python
x = 5
y = 3

print("Equal to (==):", x == y)
print("Not equal to (!=):", x != y)
print("Greater than (>):", x > y)
print("Less than (<):", x < y)
print("Greater than or equal to (>=):", x >= 5)
print("Less than or equal to (<=):", y <= 2)
````

## 4. Logical Operators
Logical operators combine conditions and evaluate boolean logic.

* `and` (Binary): Returns `True` **only if all** conditions are `True`.
* `or` (Binary): Returns `True` if **at least one** condition is `True`.
* `not` (Unary): Reverses the boolean state (`True` $\rightarrow$ `False`, `False` $\rightarrow$ `True`).

````python
x = (2 > 1)   # True
y = (1 > 2)   # False

# Logical AND (Strict)
print("x and y:", x and y)       # True and False -> False

# Logical OR (Relaxed)
print("x or y:", x or y)         # True or False -> True

# Logical NOT (Switch-flipper)
print("not x:", not x)           # not True -> False
print("not y:", not y)           # not False -> True
````

## 5. Bitwise Operators
Bitwise operators convert decimal values to binary digits (bits: `0` and `1`), align them column by column, perform bit-level logic, and convert the result back to decimal.

| Operator | Name | Logic Rule per Column |
| :--- | :--- | :--- |
| `&` | Bitwise AND | `1` only if **both** bits are `1` |
| `\|` | Bitwise OR | `1` if **at least one** bit is `1` |
| `^` | Bitwise XOR | `1` if bits are **different** (`1` and `0`) |
| `~` | Bitwise NOT | Flips all bits (`~x = -(x + 1)`) |
| `<<` | Left Shift | Shifts bits left by $n$ places (multiplies by $2^n$) |
| `>>` | Right Shift | Shifts bits right by $n$ places (integer division by $2^n$) |

#### Bit Alignment Breakdown (5 & 3):
```text
  5  ->  0 1 0 1
  3  ->  0 0 1 1
-----------------
AND  ->  0 0 0 1  (Decimal: 1)
 OR  ->  0 1 1 1  (Decimal: 7)
XOR  ->  0 1 1 0  (Decimal: 6)

````python
a = 5   # Binary: 0101
b = 3   # Binary: 0011

print("Bitwise AND (5 & 3):", a & b)      # 0001 -> 1
print("Bitwise OR  (5 | 3):", a | b)      # 0111 -> 7
print("Bitwise XOR (5 ^ 3):", a ^ b)      # 0110 -> 6
print("Bitwise NOT (~5):", ~a)            # -(5 + 1) -> -6

# Bit Shifts
print("Left Shift  (5 << 1):", 5 << 1)    # 0101 -> 01010 (10)
print("Left Shift  (5 << 2):", 5 << 2)    # 0101 -> 010100 (20)
print("Right Shift (5 >> 1):", 5 >> 1)    # 0101 -> 0010 (2)
````

## 6. Assignment & Compound Assignment Operators
The assignment operator (`=`) evaluates the expression on the right and stores it in the variable on the left.

**Compound operators** combine a binary operation with assignment:

| Compound Syntax | Equivalent Expression |
| :--- | :--- |
| `x += 2` | `x = x + 2` |
| `x -= 2` | `x = x - 2` |
| `x *= 2` | `x = x * 2` |
| `x /= 2` | `x = x / 2` |
| `x //= 2` | `x = x // 2` |
| `x %= 2` | `x = x % 2` |
| `x **= 2` | `x = x ** 2` |
| `x &= 2` | `x = x & 2` |
| `x \|= 2` | `x = x \| 2` |
| `x ^= 2` | `x = x ^ 2` |
| `x <<= 2` | `x = x << 2` |
| `x >>= 2` | `x = x >> 2` |

> **Note:** Direct assignment statements cannot be executed inside a `print()` call (e.g., `print(x = 5)` raises a `TypeError`).

````python
x = 10

x += 5     # x = 10 + 5
print("After += 5 :", x)

x *= 2     # x = 15 * 2
print("After *= 2 :", x)

x //= 4    # x = 30 // 4
print("After //= 4:", x)

x **= 2    # x = 7 ** 2
print("After **= 2:", x)
````

## 7. Identity Operators (`is`, `is not`)
Identity operators check whether two variables refer to the **exact same memory location (address in RAM)**, not just whether their values are equal.

* `is`: Returns `True` if `id(a) == id(b)`
* `is not`: Returns `True` if `id(a) != id(b)`

> **Memory Rule:** Immutable types with the same value (like small integers and short strings) often share the same memory location due to interning. Mutable types (like lists and dictionaries) receive distinct memory addresses even if their contents are identical.

````python
# Immutable objects (integers)
a = 2
b = 2
print("a is b (integers):", a is b)          # True (same RAM address)
print("id(a) == id(b):", id(a) == id(b))

# Mutable objects (lists)
list1 = [1, 2, 3]
list2 = [1, 2, 3]
print("list1 == list2 (Value check):", list1 == list2)   # True
print("list1 is list2 (Memory check):", list1 is list2)  # False (different RAM addresses)
print("list1 is not list2:", list1 is not list2)         # True
````

## 8. Membership Operators (`in`, `not in`)
Membership operators test whether a specific element exists within a sequence or container collection (e.g., `str`, `list`, `tuple`, `set`, `dict`).

* `in`: Returns `True` if the item is found inside the container.
* `not in`: Returns `True` if the item is **not** found inside the container.

````python
city = "Peshawar"
numbers = [10, 20, 30, 40, 50]

# Testing strings
print("'P' in 'Peshawar':", "P" in city)       # True
print("'z' in 'Peshawar':", "z" in city)       # False

# Testing collections
print("20 in numbers:", 20 in numbers)         # True
print("100 not in numbers:", 100 not in numbers) # True
````
