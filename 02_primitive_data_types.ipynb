# Day 3: The Python Data Types Blueprint 🏗️

In Python, all data is categorized into specific types so the computer knows how to store, display, and perform operations on it.

### 🗺️ The Three Core Families of Data Types
Before we look at the values, we must understand how Python organizes data types into three main categories:

1. **Primitive (Basic) Types:** Single, simple, atomic values that cannot be broken down into smaller, independent data types (e.g., Numbers, Text, Booleans). **This is our focus for today!**
2. **Container (Collection) Types:** Structures used to group and store multiple values together (e.g., Lists, Tuples, Dictionaries, Sets).
3. **User-Defined Types (Classes):** Custom blueprints created by programmers to represent real-world objects.

---

## Part 1: The 5 Primitive Pillars

Today, we will learn about the 5 primitive types by looking directly at how Python handles their raw values. We will not use variables today, as they are a topic for another day!

### 1. Integers (`int`)
* Represents positive, negative, and zero whole numbers without decimals.
* **Large Numbers:** Do not use commas or decimals in large numbers. Instead, use underscores (`_`) to make them readable (e.g., `1_000` instead of `1,000`).
* **Division vs. Floor Division:** * Normal division (`/`) of two integers *always* outputs a **Float** (e.g., `2 / 3` is `0.6666...`).
  * Floor division (`//`) performs the calculation and drops everything after the decimal, outputting only the whole number before the decimal (e.g., `2 // 3` becomes `0`).
* **Size Limit:** The largest integer Python can represent is $10^{308}$ (written as `1e308`). Anything at `1e309` or higher is treated as infinity (`inf`).

Let's test these integer rules in the code cell below:

```python
# Displaying integers and their types
print("Raw Integer:    ", 42, " | Type: ", type(42))
print("Negative Integer:", -15, " | Type: ", type(-15))
print("Using Underscores:", 1_000_000)

# Normal Division vs Floor Division
print("Normal Division (2 / 3):", 2 / 3, " | Type:", type(2 / 3))
print("Floor Division (2 // 3):", 2 // 3, " | Type:", type(2 // 3))

# Integer Limits
print("Maximum Int (1e308):", 1e308)
print("Overflow (1e309):   ", 1e309)
```

### 2. Floats (`float`)
* Python's special name for numbers with a decimal point (e.g., `5.0`, `6.999`).
* Do not use quotes or commas. A decimal point is mandatory.
* **Mixed Operations:** Any mathematical operation (`+`, `-`, `*`, `/`) between a float and an integer will automatically output a **Float**.
* **Size Limit:** The largest float you can write in Python is `1.7e308`.

Let's test these float rules:

```python
# Displaying floats and their types
print("Raw Float:       ", 5.0, " | Type: ", type(5.0))
print("Float arithmetic:", 5.0 * 2, " | Type: ", type(5.0 * 2))

# Float Limit
print("Max Float (1.7e308):", 1.7e308)
print("Float Overflow:     ", 1.8e308)
```

### 3. Complex Numbers (`complex`)
* Consists of two parts: a **real part** and an **imaginary part**.
* In mathematics, engineers use $i$ for the imaginary unit ($\sqrt{-1}$), but Python uses **`j`** or **`J`**.
* **Syntax Rule:** There must be absolutely **no space** between the number and the `j` (e.g., `1j`, `2.5j`). If you write `j` by itself, Python will think it is a letter/variable and crash. Always write `1j` if you mean just the imaginary unit.
* **Extracting Parts:** You can extract the real and imaginary parts directly from a complex number using `.real` and `.imag`. When doing this with raw numbers, wrap the complex number in parentheses first (e.g., `(3 + 4j).real`). Note that Python always returns these extracted parts as **Floats**.
* **Operations & Conjugation:** You can perform math on complex numbers easily.
  * Multiplying two imaginary units ($1j \times 1j$) results in $-1$ (specifically, `-1+0j` in Python).
  * You can find the complex conjugate (flipping the sign of the imaginary part) by running `.conjugate()` directly on the complex number enclosed in parentheses.

```python
# Displaying complex numbers and their types
print("Complex Value:", 3 + 4j, " | Type: ", type(3 + 4j))

# Extracting real and imaginary parts without variables
print("Real Part:     ", (3 + 4j).real)
print("Imaginary Part:", (3 + 4j).imag)
print("Both Parts:    ", (3 + 4j).real, (3 + 4j).imag)

# Multiplying imaginary units (j * j = -1)
print("Multiplying 1j * 1j:", 1j * 1j)

# Finding the Conjugate of a complex number
print("Conjugate of (3 + 4j):", (3 + 4j).conjugate())
```

### 4. Booleans (`bool`)
* A logical data type that can only ever have one of two values: **`True`** or **`False`**.
* **Syntax Rule:** Never wrap them in quotations, and the first letter must **always be Capital** (`True` and `False`).
* **Under the Hood:** Behind the scenes, Python treats `True` as `1` and `False` as `0`. You can actually use them in arithmetic!

```python
# Displaying booleans and their types
print("Boolean True: ", True, " | Type: ", type(True))
print("Boolean False:", False, " | Type: ", type(False))

# Under the hood behavior (Arithmetic with Booleans)
print("True + True (1 + 1):  ", True + True)
print("False + True (0 + 1): ", False + True)
print("True * 50 (1 * 50):   ", True * 50)
```

### 5. Strings (`str`)
* Represents plain text. You must wrap the text inside single (`'`) or double (`"`) quotes.
* When Python sees quotations, it stops trying to calculate math and simply displays exactly what is typed inside.
* **Multi-line Strings:** If your text spans across multiple lines, wrap it inside triple quotations (`"""` or `'''`).
* **String Addition (Concatenation):** You can glue strings together using the `+` operator. To add a space between two words, you must explicitly glue a space string `" "` between them.

```python
# Displaying strings and their types
print("Double Quotes String:", "Hello Python!", " | Type: ", type("Hello Python!"))
print("Single Quotes String:", 'Day 3 Study Notes')

# Multi-line string displaying
print("""
This is line 1.
This is line 2.
This is line 3.
""")

# String Addition (Concatenation) without variables
print("Muhamad" + " " + "Shahan")
```
