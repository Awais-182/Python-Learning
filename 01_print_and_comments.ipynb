## 1. Writing Comments in Python
Comments are lines of text in your code file that Python completely ignores. They are used to document your code, explain complex logic, or temporarily disable code lines during testing.

* **Single-line comments:** Created by putting a hash symbol (`#`) at the start of the line.
* **Multi-line comments:** Created by wrapping your text inside triple quotes (`"""` or `'''`).

```python
# This is a single-line comment. Python completely ignores this line!

print("Comments are awesome!") # You can also write a comment at the end of a code line

"""
This is a multi-line comment.
I can write as many lines as I want here
to explain my program in detail.
Python will ignore all of this too!
"""
print("Multi-line comments are clean and professional.")
```

## 2. Basic Printing
We can use the `print()` function to output text, numbers, or even mathematical calculations.

* **Rule:** Text (known as a string) **must** be wrapped in quotation marks. Numbers and calculations must **not** have quotation marks, or Python will treat them as plain text instead of calculating them.

```python
# 1. Printing Text (Requires Quotation Marks!)
print("Hello world!")

# 2. Printing Whole Numbers (No quotes needed!)
print(16)

# 3. Printing Math Calculations (Python does the math first, then prints the result!)
print(5 + 5)
```

## 3. Printing Multiple Values in One Go
Instead of writing multiple `print()` commands on separate lines, you can display multiple items in a single line by separating them with a **comma (`,`)**.

By default, Python is smart enough to automatically put a **single space** between each item.

```python
# Creating variables
name = "Awais"
age = 16

# Printing variables and raw text together using commas
print("My name is", name, "and I am", age, "years old.")
```

## 4. Customizing the Spacer: The `sep` Parameter
What if you don't want a default space between your printed items? What if you want a hyphen (`-`), a slash (`/`), or nothing at all?

We use the **`sep`** (short for separator) parameter at the end of our print function to change this behavior.

```python
# 1. Default spacing (Standard space)
print("Apple", "Banana", "Cherry")

# 2. Hyphen separation
print("Apple", "Banana", "Cherry", sep="-")

# 3. Slash separation (Great for writing dates!)
print("15", "07", "2026", sep="/")
```

## 5. Keeping Things on the Same Line: The `end` Parameter
Normally, every time you use `print()`, Python automatically hits "Enter" and moves to a **new line** for the next print statement.

If you want to prevent this newline, or print a special symbol at the very end of your print statement, you can use the **`end`** parameter.

```python
# 1. Normal behavior (Prints on separate lines)
print("Loading")
print("Complete!")

print("-" * 15)  # Quick divider line using string repetition

# 2. Using 'end' to keep the next print on the same line!
print("Loading", end="... ")
print("Complete!")
```
